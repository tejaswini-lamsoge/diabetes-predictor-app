```python
# ============================================
# STEP: CLEAN AND PREPARE THE DATASET
# ============================================

import pandas as pd

# 1. Start from the original dataset
clean_df = df.copy()

# 2. Remove exact duplicate rows
clean_df = clean_df.drop_duplicates().reset_index(drop=True)

# 3. Columns containing Yes/No values
yes_no_columns = [
    "Polyuria",
    "Polydipsia",
    "sudden weight loss",
    "weakness",
    "Polyphagia",
    "Genital thrush",
    "visual blurring",
    "Itching",
    "Irritability",
    "delayed healing",
    "partial paresis",
    "muscle stiffness",
    "Alopecia",
    "Obesity"
]

# 4. Convert Yes/No to 1/0
for column in yes_no_columns:
    clean_df[column] = clean_df[column].map({
        "Yes": 1,
        "No": 0
    })

# 5. Convert Gender to 1/0
clean_df["Gender"] = clean_df["Gender"].map({
    "Male": 1,
    "Female": 0
})

# 6. Convert target class to 1/0
clean_df["class"] = clean_df["class"].map({
    "Positive": 1,
    "Negative": 0
})

# 7. Check for missing values created during conversion
print("Missing values:")
print(clean_df.isnull().sum())

# 8. Check data types
print("\nData types:")
print(clean_df.dtypes)

# 9. Create X (features) and y (target)
X = clean_df.drop("class", axis=1)
y = clean_df["class"]

# 10. Print final dataset information
print("\nFinal dataset shape:")
print("X shape:", X.shape)
print("y shape:", y.shape)

# 11. Display first 5 rows
print("\nFirst 5 rows of cleaned dataset:")
display(clean_df.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

#training the model starts

#Linear_regression_model

```python
# ============================================
# FIRST MACHINE LEARNING MODEL
# Logistic Regression
# ============================================

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# 1. Create model
model = LogisticRegression(max_iter=1000)

# 2. Train model
model.fit(X_train, y_train)

print("Model training completed!")

# 3. Make predictions on unseen test data
y_pred = model.predict(X_test)

# 4. Show predictions
print("\nPredictions:")
print(y_pred)

# 5. Show actual answers
print("\nActual:")
print(y_test.to_numpy())

# 6. Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)

# 7. Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 8. Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Models
logistic_model = LogisticRegression(max_iter=1000)

tree_model = DecisionTreeClassifier(
    random_state=42
)

forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Cross-validation scores
logistic_scores = cross_val_score(
    logistic_model, X_train, y_train, cv=cv, scoring="accuracy"
)

tree_scores = cross_val_score(
    tree_model, X_train, y_train, cv=cv, scoring="accuracy"
)

forest_scores = cross_val_score(
    forest_model, X_train, y_train, cv=cv, scoring="accuracy"
)

# Print results
print("Logistic Regression:")
print(logistic_scores)
print("Mean:", logistic_scores.mean())

print("\nDecision Tree:")
print(tree_scores)
print("Mean:", tree_scores.mean())

print("\nRandom Forest:")
print(forest_scores)
print("Mean:", forest_scores.mean())
```

```python
# ============================================
# RANDOM FOREST HYPERPARAMETER TUNING
# ============================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold

# 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Base Random Forest
rf = RandomForestClassifier(
    random_state=42
)

# Hyperparameters we want to test
param_grid = {
    "n_estimators": [50, 100, 200],
    "max_depth": [None, 5, 10, 15],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4]
}

# Grid search
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

# Train and test all combinations
grid_search.fit(X_train, y_train)

# Best parameters
print("Best Parameters:")
print(grid_search.best_params_)

# Best cross-validation score
print("\nBest Cross-Validation Accuracy:")
print(grid_search.best_score_)

print("\nBest Cross-Validation Accuracy (%):")
print(grid_search.best_score_ * 100)
```

```python
# ============================================
# FINAL TEST OF TUNED RANDOM FOREST
# ============================================

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Get the best model found during tuning
best_rf = grid_search.best_estimator_

# Predict on the untouched test data
final_pred = best_rf.predict(X_test)

# Calculate final accuracy
final_accuracy = accuracy_score(y_test, final_pred)

print("Final Test Accuracy:")
print(final_accuracy)

print("\nFinal Test Accuracy (%):")
print(final_accuracy * 100)

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, final_pred))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, final_pred))
```

```python
# ============================================
# FEATURE IMPORTANCE
# ============================================

import pandas as pd

# Get feature importance from the tuned Random Forest
importance = best_rf.feature_importances_

# Create a table
feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": importance
})

# Sort from highest to lowest
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)
```

```python
# ============================================
#create prediction function
# ============================================

def predict_diabetes(
    age,
    gender,
    polyuria,
    polydipsia,
    sudden_weight_loss,
    weakness,
    polyphagia,
    genital_thrush,
    visual_blurring,
    itching,
    irritability,
    delayed_healing,
    partial_paresis,
    muscle_stiffness,
    alopecia,
    obesity
):

    # Convert user input into the same numerical format
    input_data = pd.DataFrame([{
        "Age": age,
        "Gender": 1 if gender == "Male" else 0,
        "Polyuria": 1 if polyuria == "Yes" else 0,
        "Polydipsia": 1 if polydipsia == "Yes" else 0,
        "sudden weight loss": 1 if sudden_weight_loss == "Yes" else 0,
        "weakness": 1 if weakness == "Yes" else 0,
        "Polyphagia": 1 if polyphagia == "Yes" else 0,
        "Genital thrush": 1 if genital_thrush == "Yes" else 0,
        "visual blurring": 1 if visual_blurring == "Yes" else 0,
        "Itching": 1 if itching == "Yes" else 0,
        "Irritability": 1 if irritability == "Yes" else 0,
        "delayed healing": 1 if delayed_healing == "Yes" else 0,
        "partial paresis": 1 if partial_paresis == "Yes" else 0,
        "muscle stiffness": 1 if muscle_stiffness == "Yes" else 0,
        "Alopecia": 1 if alopecia == "Yes" else 0,
        "Obesity": 1 if obesity == "Yes" else 0
    }])

    # Make prediction
    prediction = best_rf.predict(input_data)[0]

    # Convert numerical prediction back to text
    if prediction == 1:
        result = "Positive"
    else:
        result = "Negative"

    return result
```

```python
#==========================
#add prediction probability
#==========================
def predict_diabetes(
    age,
    gender,
    polyuria,
    polydipsia,
    sudden_weight_loss,
    weakness,
    polyphagia,
    genital_thrush,
    visual_blurring,
    itching,
    irritability,
    delayed_healing,
    partial_paresis,
    muscle_stiffness,
    alopecia,
    obesity
):

    input_data = pd.DataFrame([{
        "Age": age,
        "Gender": 1 if gender == "Male" else 0,
        "Polyuria": 1 if polyuria == "Yes" else 0,
        "Polydipsia": 1 if polydipsia == "Yes" else 0,
        "sudden weight loss": 1 if sudden_weight_loss == "Yes" else 0,
        "weakness": 1 if weakness == "Yes" else 0,
        "Polyphagia": 1 if polyphagia == "Yes" else 0,
        "Genital thrush": 1 if genital_thrush == "Yes" else 0,
        "visual blurring": 1 if visual_blurring == "Yes" else 0,
        "Itching": 1 if itching == "Yes" else 0,
        "Irritability": 1 if irritability == "Yes" else 0,
        "delayed healing": 1 if delayed_healing == "Yes" else 0,
        "partial paresis": 1 if partial_paresis == "Yes" else 0,
        "muscle stiffness": 1 if muscle_stiffness == "Yes" else 0,
        "Alopecia": 1 if alopecia == "Yes" else 0,
        "Obesity": 1 if obesity == "Yes" else 0
    }])

    prediction = best_rf.predict(input_data)[0]

    probabilities = best_rf.predict_proba(input_data)[0]

    negative_probability = probabilities[0]
    positive_probability = probabilities[1]

    if prediction == 1:
        result = "Positive"
    else:
        result = "Negative"

    return result, negative_probability, positive_probability
```

```python
result, negative_prob, positive_prob = predict_diabetes(
    age=45,
    gender="Male",
    polyuria="Yes",
    polydipsia="Yes",
    sudden_weight_loss="Yes",
    weakness="Yes",
    polyphagia="Yes",
    genital_thrush="No",
    visual_blurring="Yes",
    itching="No",
    irritability="Yes",
    delayed_healing="Yes",
    partial_paresis="No",
    muscle_stiffness="No",
    alopecia="No",
    obesity="Yes"
)

print("Prediction:", result)
print("Negative probability:", round(negative_prob * 100, 2), "%")
print("Positive probability:", round(positive_prob * 100, 2), "%")
```
